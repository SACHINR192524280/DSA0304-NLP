```python
import re

resume = """
Name: Rahul Kumar
Email: rahul.kumar@gmail.com
Mobile: 9876543210
Skills: Python, Java, SQL, Machine Learning, NLP
Experience: 3 years
"""

# Extract name
name = re.search(r"Name:\s*([A-Za-z ]+)", resume)

# Extract email
email = re.search(r"[\w.-]+@[\w.-]+\.\w+", resume)

# Extract mobile number
mobile = re.search(r"\b[6-9]\d{9}\b", resume)

# Extract technical skills
skills_list = ["Python", "Java", "SQL", "Machine Learning", "NLP"]
skills = []

for skill in skills_list:
    if re.search(r"\b" + re.escape(skill) + r"\b", resume, re.IGNORECASE):
        skills.append(skill)

# Extract experience
experience = re.search(r"Experience:\s*(\d+)\s*years?", resume)

# Display extracted information
candidate_name = name.group(1).strip() if name else "Not Found"
candidate_email = email.group() if email else "Not Found"
candidate_mobile = mobile.group() if mobile else "Not Found"
candidate_experience = int(experience.group(1)) if experience else 0

print("===== CANDIDATE PROFILE =====")
print("Name:", candidate_name)
print("Email:", candidate_email)
print("Mobile:", candidate_mobile)
print("Skills:", ", ".join(skills))
print("Experience:", candidate_experience, "years")

print("\n===== PROFILE SUMMARY =====")
print(
    f"{candidate_name} has {candidate_experience} years of experience "
    f"with skills in {', '.join(skills)}."
)

# Eligibility check
if candidate_experience >= 2 and any(
    skill.lower() == "python" for skill in skills
):
    print("\nEligibility: ELIGIBLE")
else:
    print("\nEligibility: NOT ELIGIBLE")
```

```python
import re

products = [
    "Apple iPhone 15",
    "Samsung Galaxy S24",
    "OnePlus 12",
    "Apple MacBook Air",
    "Dell Laptop",
    "HP Laptop",
    "Samsung Galaxy Tab",
    "Sony Headphones",
    "Apple Watch"
]

def search_products(keyword, search_type):
    matches = []

    for product in products:

        if search_type == "exact":
            pattern = r"\b" + re.escape(keyword) + r"\b"

        elif search_type == "prefix":
            pattern = r"\b" + re.escape(keyword)

        elif search_type == "suffix":
            pattern = re.escape(keyword) + r"\b"

        elif search_type == "partial":
            pattern = re.escape(keyword)

        else:
            return []

        if re.search(pattern, product, re.IGNORECASE):
            matches.append(product)

    return matches


print("===== PRODUCT SEARCH SYSTEM =====")
keyword = input("Enter search keyword: ")

print("\n1. Exact Search")
print("2. Prefix Search")
print("3. Suffix Search")
print("4. Partial Search")

choice = input("Enter choice: ")

search_type = {
    "1": "exact",
    "2": "prefix",
    "3": "suffix",
    "4": "partial"
}.get(choice)

if search_type:
    results = search_products(keyword, search_type)

    print("\nMatching Products:")

    if results:
        for product in results:
            print("-", product)

    else:
        print("No products found.")

    print("\nTotal Matching Products:", len(results))

else:
    print("Invalid choice.")
```

```python
import re

print("===== UNIVERSITY REGISTRATION SYSTEM =====")

register_no = input("Enter Register Number: ")
email = input("Enter Institutional Email: ")
course_code = input("Enter Course Code: ")
semester = input("Enter Semester: ")
mobile = input("Enter Mobile Number: ")

# Validation patterns
register_pattern = r"^\d{9}$"
email_pattern = r"^[A-Za-z0-9._%+-]+@saveetha\.com$"
course_pattern = r"^[A-Z]{3}\d{4}$"
semester_pattern = r"^(1|2|3|4|5|6|7|8)$"
mobile_pattern = r"^[6-9]\d{9}$"

# Validate Register Number
if re.fullmatch(register_pattern, register_no):
    print("Register Number: Valid")
    reg_valid = True
else:
    print("Register Number: Invalid")
    reg_valid = False

# Validate Email
if re.fullmatch(email_pattern, email):
    print("Institutional Email: Valid")
    email_valid = True
else:
    print("Institutional Email: Invalid")
    email_valid = False

# Validate Course Code
if re.fullmatch(course_pattern, course_code):
    print("Course Code: Valid")
    course_valid = True
else:
    print("Course Code: Invalid")
    course_valid = False

# Validate Semester
if re.fullmatch(semester_pattern, semester):
    print("Semester: Valid")
    semester_valid = True
else:
    print("Semester: Invalid")
    semester_valid = False

# Validate Mobile Number
if re.fullmatch(mobile_pattern, mobile):
    print("Mobile Number: Valid")
    mobile_valid = True
else:
    print("Mobile Number: Invalid")
    mobile_valid = False

# Final status
if all([
    reg_valid,
    email_valid,
    course_valid,
    semester_valid,
    mobile_valid
]):
    print("\nFinal Registration Status: SUCCESSFUL")
else:
    print("\nFinal Registration Status: FAILED")
```
