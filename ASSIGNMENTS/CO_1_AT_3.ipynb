```python
import re

def validate_email(email):
    pattern = r'^[A-Za-z][A-Za-z0-9._]*@[A-Za-z]+\.(com|org|edu|net|in)$'
    return bool(re.fullmatch(pattern, email))


def validate_password(password):
    pattern = r'^(?=.*[A-Z])(?=.*[a-z])(?=.*\d)(?=.*[@#$%&!]).{8,}$'
    return bool(re.fullmatch(pattern, password))


def validate_mobile(mobile):
    pattern = r'^[6-9]\d{9}$'
    return bool(re.fullmatch(pattern, mobile))


print("===== INTELLIGENT VALIDATOR =====")

email = input("Enter Email: ")
password = input("Enter Password: ")
mobile = input("Enter Mobile Number: ")

print("\n===== VALIDATION RESULTS =====")

if validate_email(email):
    print("Email: Valid Email")
else:
    print("Email: Invalid Email")

if validate_password(password):
    print("Password: Strong Password")
else:
    print("Password: Weak Password")

if validate_mobile(mobile):
    print("Mobile: Valid Mobile Number")
else:
    print("Mobile: Invalid Mobile Number")
```

```python
def simulate_dfa(string):
    transitions = {
        "q0": {"a": "q1", "b": "q0"},
        "q1": {"a": "q1", "b": "q2"},
        "q2": {"a": "q1", "b": "q0"}
    }

    current_state = "q0"
    path = [current_state]

    for symbol in string:

        if symbol not in ["a", "b"]:
            return None, path

        current_state = transitions[current_state][symbol]
        path.append(current_state)

    return current_state, path


print("===== DFA SIMULATOR =====")
print("Language: Strings ending with 'ab'")

n = int(input("Enter number of strings: "))

for i in range(n):

    string = input(f"\nEnter string {i + 1}: ")

    final_state, path = simulate_dfa(string)

    if final_state is None:
        print("Invalid input. Use only a and b.")
        continue

    print("Transition Path:", " -> ".join(path))

    if final_state == "q2":
        print("Accepted")
    else:
        print("Rejected")
```

```python
import re

text = """
Meeting on 12/09/2026
Call 9876543210
#NLP
@OpenAI
natural language processing
Python programming
"""

def search_date():
    pattern = r'\b\d{2}/\d{2}/\d{4}\b'
    return re.findall(pattern, text)


def search_phone():
    pattern = r'\b[6-9]\d{9}\b'
    return re.findall(pattern, text)


def search_hashtag():
    pattern = r'#\w+'
    return re.findall(pattern, text)


def search_mention():
    pattern = r'@\w+'
    return re.findall(pattern, text)


def search_prefix(prefix):
    pattern = r'\b' + re.escape(prefix) + r'\w*'
    return re.findall(pattern, text, re.IGNORECASE)


def search_suffix(suffix):
    pattern = r'\b\w*' + re.escape(suffix) + r'\b'
    return re.findall(pattern, text, re.IGNORECASE)


def search_word(word):
    pattern = r'\b' + re.escape(word) + r'\b'
    return re.findall(pattern, text, re.IGNORECASE)


while True:

    print("\n===== SMART PATTERN MATCHING ENGINE =====")
    print("1. Search Date")
    print("2. Search Phone Number")
    print("3. Search Hashtag")
    print("4. Search Mention")
    print("5. Search Prefix")
    print("6. Search Suffix")
    print("7. Search Word")
    print("8. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        print("Dates:", search_date())

    elif choice == "2":
        print("Phone Numbers:", search_phone())

    elif choice == "3":
        print("Hashtags:", search_hashtag())

    elif choice == "4":
        print("Mentions:", search_mention())

    elif choice == "5":
        prefix = input("Enter prefix: ")
        print("Prefix Matches:", search_prefix(prefix))

    elif choice == "6":
        suffix = input("Enter suffix: ")
        print("Suffix Matches:", search_suffix(suffix))

    elif choice == "7":
        word = input("Enter word: ")
        print("Word Matches:", search_word(word))

    elif choice == "8":
        print("Exiting...")
        break

    else:
        print("Invalid choice.")
```
