```python
# 1.question?

from nltk.stem import PorterStemmer, WordNetLemmatizer

ps = PorterStemmer()
lem = WordNetLemmatizer()

words = ["analyzing", "analysis", "analytical"]

print("Word\t\tRoot\t\tAffix\tType\t\tNormalized")

for word in words:
    if word == "analyzing":
        root, affix, category, norm = "analyze", "ing", "Inflectional", "analyze"
    elif word == "analysis":
        root, affix, category, norm = "analysis", "-", "Derivational", "analyze"
    else:
        root, affix, category, norm = "analyt", "ical", "Derivational", "analyze"

    print(f"{word}\t\t{root}\t\t{affix}\t{category}\t{norm}")
    print("NLTK Stem:", ps.stem(word))
```

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer

ps = PorterStemmer()
lem = WordNetLemmatizer()

words = ["disagree", "agreement", "agreeable"]

rules = {
    "disagree": ("dis", "agree", "-", "Derivational",
                 "Opposite or contrary meaning"),
    "agreement": ("-", "agree", "ment", "Derivational",
                  "State or result of agreeing"),
    "agreeable": ("-", "agree", "able", "Derivational",
                  "Capable of being pleasing")
}

print("Word\t\tPrefix\tRoot\tSuffix\tType\t\tMeaning")

for word in words:
    prefix, root, suffix, category, meaning = rules[word]
    print(f"{word}\t{prefix}\t{root}\t{suffix}\t{category}\t{meaning}")
    print("NLTK Stem:", ps.stem(word))
```

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer

ps = PorterStemmer()
lem = WordNetLemmatizer()

words = ["govern", "government", "governance"]

rules = {
    "govern": ("govern", "-", 0),
    "government": ("govern", "ment", 1),
    "governance": ("govern", "ance", 1)
}

print("Word\t\tRoot\t\tAffix\tLevel\tNormalized")

for word in words:
    root, affix, level = rules[word]
    print(f"{word}\t\t{root}\t\t{affix}\t{level}\tgovern")
    print("NLTK Stem:", ps.stem(word))
```

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer

ps = PorterStemmer()
lem = WordNetLemmatizer()

words = ["activate", "activation", "reactivation"]

rules = {
    "activate": ("-", "act", "ivate", "Verb", "Base verb"),
    "activation": ("-", "act", "ivation", "Noun", "Verb to noun"),
    "reactivation": ("re", "act", "ivation", "Noun",
                     "Prefix re- indicates repetition")
}

print("Word\t\tPrefix\tRoot\tSuffix\t\tClass")

for word in words:
    prefix, root, suffix, word_class, meaning = rules[word]
    print(f"{word}\t{prefix}\t{root}\t{suffix}\t{word_class}")
    print("Derivation:", meaning)
    print("NLTK Stem:", ps.stem(word))
```

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer

ps = PorterStemmer()
lem = WordNetLemmatizer()

words = ["create", "creates", "creating"]

rules = {
    "create": ("-", "Base form"),
    "creates": ("s", "Third-person singular"),
    "creating": ("ing", "Present participle")
}

print("Word\t\tSuffix\tCategory\t\tStem\tNormalized")

for word in words:
    suffix, category = rules[word]
    stem = ps.stem(word)
    normalized = lem.lemmatize(word, pos="v")
    print(f"{word}\t\t{suffix}\t{category}\t{stem}\t{normalized}")
```
